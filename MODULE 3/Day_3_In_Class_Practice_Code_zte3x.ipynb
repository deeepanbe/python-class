```python
# Fucntion way of coding 

def <Name_of_the_Fucntion>([list of arguments]):
    <Block of Statements>
    return <Value >
```

```python
# Step 1 Function Defintion
def simpleInterest(P, N, R):
    SI = (P * N * R)/100
    
    return SI 
```

```python
# Step2 Function calling part
# Customer 1   
P = 1000
 
N = 5
 
R = 7.5
output = simpleInterest(P, N, R)
#Step 3  result 
print('Customer 1 SI =', output)
```

```python
# Customer 2   
P = 3000
 
N = 10
 
R = 6.7
output = simpleInterest(P, N, R)
print('Customer 2 SI =', output)
```

```python
class Fruit:
    # Constructor , get called automatically at the time of object creation
    def __init__(self, name):
        self.name = name
        print('Constructor get called')
    # Member Method
    def display(self):
        print('Name of the Fruit = ',self.name)
```
