```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 1: Extract data into NumPy arrays
def extract_data(file_path):
    data = pd.read_csv(file_path)
    salary = data['Salary'].to_numpy()
    gender = data['Gender'].to_numpy()
    age = data['Age'].to_numpy()
    phd = data['PhD'].to_numpy()
    return salary, gender, age, phd

# Task 2: Count men and women with a PhD
def count_phd(gender, phd):
    men_phd = np.sum((gender == 'Male') & (phd == 1))
    women_phd = np.sum((gender == 'Female') & (phd == 1))
    return men_phd, women_phd

# Task 3: Filter data with PhD
def filter_phd(data):
    return data[data['PhD'] == 1]

# Task 4: Total number of PhD holders
def total_phd(phd):
    return np.sum(phd == 1)

# Task 5: Count occurrences of values in an array
def count_occurrences(arr):
    return np.bincount(arr)

# Task 6: Filter elements greater than 5
def filter_greater_than_five(arr):
    return arr[arr > 5]

# Task 7: Handle NaN in arrays
def handle_nan():
    arr = np.array([np.nan, 1, 2, np.nan, 3, 4, 5])
    return arr[~np.isnan(arr)]

# Task 8: Min and max in a 10x10 array
def min_max_10x10():
    arr = np.random.random((10, 10))
    return np.min(arr), np.max(arr)

# Task 9: Mean of a random vector
def mean_random_vector():
    vec = np.random.random(30)
    return np.mean(vec)

# Task 10: Negate elements between 3 and 9
def negate_elements():
    arr = np.arange(11)
    arr[(arr >= 3) & (arr <= 9)] *= -1
    return arr

# Task 11: Sort 3x3 array by columns
def sort_array():
    arr = np.random.randint(0, 10, (3, 3))
    return np.sort(arr, axis=0)

# Task 12: Sum over last two axes
def sum_last_two_axes():
    arr = np.random.random((3, 4, 4))
    return np.sum(arr, axis=(-2, -1))

# Task 13: Swap rows in an array
def swap_rows():
    arr = np.random.randint(0, 10, (3, 3))
    arr[[0, 1]] = arr[[1, 0]]
    return arr

# Task 14: Compute matrix rank
def compute_rank():
    mat = np.random.randint(0, 10, (3, 3))
    return np.linalg.matrix_rank(mat)

# Task 15: Analyze Tennessee school outcomes
def analyze_school_data(file_path):
    data = pd.read_csv(file_path)

    # Phase 1: Describe the data
    print("Data Description:\n", data.describe())

    # Phase 2: Group by school rating
    grouped = data.groupby('school_rating')['reduced_lunch'].describe()
    print("Grouped Data:\n", grouped)

    # Phase 3: Correlation analysis
    correlation = data[['reduced_lunch', 'school_rating']].corr()
    print("Correlation Matrix:\n", correlation)

    # Phase 4: Scatter Plot
    plt.scatter(data['reduced_lunch'], data['school_rating'])
    plt.xlabel('Reduced Lunch')
    plt.ylabel('School Rating')
    plt.title('Scatter Plot of Reduced Lunch vs School Rating')
    plt.show()

    # Phase 5: Correlation Matrix Heatmap
    sns.heatmap(correlation, annot=True, cmap='coolwarm')
    plt.title('Correlation Matrix Heatmap')
    plt.show()

```
