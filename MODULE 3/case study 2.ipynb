```python
import csv

def load_data(file_path):
    """Reads the CSV file and extracts professions and ages."""
    professions = set()
    ages = []

    try:
        with open(file_path, mode='r') as file:
            csv_reader = csv.DictReader(file)
            for row in csv_reader:
                professions.add(row['job'].strip().lower())  # Make profession check case-insensitive
                ages.append(int(row['age']))
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None, None

    return professions, ages

def check_eligibility(professions, ages):
    """Checks eligibility based on profession and age."""
    if not professions or not ages:
        print("No data to process.")
        return

    max_age = max(ages)
    min_age = min(ages)

    print(f"Max age for eligibility: {max_age}")
    print(f"Min age for eligibility: {min_age}")
    print("Type 'END' to exit.")

    while True:
        profession = input("Enter the profession to check eligibility: ").strip().lower()

        if profession == "end":
            print("Exiting the program.")
            break

        if profession in professions:
            print("The client is eligible for the marketing campaign.")
        else:
            print("The client is not eligible for the marketing campaign.")

def main():
    # Use a raw string to avoid issues with backslashes in the file path
    file_path = r"C:\Users\dell\Desktop\EDUREKA\PYTHON\MODULE 3\bank-data.csv"
    professions, ages = load_data(file_path)
    check_eligibility(professions, ages)

if __name__ == "__main__":
    main()
```
