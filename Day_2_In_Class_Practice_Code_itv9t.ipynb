```python
day = 5
match day:
    case 1:
        print('Sun')
    case 2:
        print('mon')        
    case 3:
        print('tue')        
    case 4:
        print('wed')        
    case 5:
        print('thu')        
    case 6:
        print('fri')        
    case 7:
        print('sat')        
    case _:
        print('Enter the valid Day number')
```
