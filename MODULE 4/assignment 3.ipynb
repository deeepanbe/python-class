```python
# Load the SalaryGender data
try:
    salary_data = pd.read_csv('SalaryGender.csv')

    # Convert columns to NumPy arrays
    salary = salary_data['Salary'].to_numpy()
    gender = salary_data['Gender'].to_numpy()
    age = salary_data['Age'].to_numpy()
    phd = salary_data['PhD'].to_numpy()

    # Count men and women with Ph.D.
    men_with_phd = salary_data[(salary_data['Gender'] == 'Male') & (salary_data['PhD'] == 1)].shape[0]
    women_with_phd = salary_data[(salary_data['Gender'] == 'Female') & (salary_data['PhD'] == 1)].shape[0]
    print(f"Men with PhD: {men_with_phd}, Women with PhD: {women_with_phd}")
except FileNotFoundError:
    print("File 'SalaryGender.csv' not found. Please ensure the dataset is available.")
```

```python
try:
    # Load datasets
    math_scores = pd.read_csv('MathScoreTerm1.csv')
    ds_scores = pd.read_csv('DSScoreTerm1.csv')
    physics_scores = pd.read_csv('PhysicsScoreTerm1.csv')

    # Remove confidential columns and fill missing scores with 0
    for df in [math_scores, ds_scores, physics_scores]:
        df.drop(['Name', 'Ethnicity'], axis=1, errors='ignore', inplace=True)
        df.fillna(0, inplace=True)

    # Merge the three datasets
    merged_scores = math_scores.merge(ds_scores, on='ID').merge(physics_scores, on='ID')
    merged_scores['Sex'] = merged_scores['Sex'].map({'M': 1, 'F': 2})
    merged_scores.to_csv('ScoreFinal.csv', index=False)
    print("Merged scores saved to 'ScoreFinal.csv'.")
except FileNotFoundError:
    print("One or more files for Case Study 2 are missing. Please check.")
```

```python
try:
    # Load hurricane data and plot bar graph
    hurricane_data = pd.read_csv('hurricane_data.csv')
    plt.bar(hurricane_data['Year'], hurricane_data['Hurricanes'])
    plt.title('Number of Hurricanes by Year')
    plt.xlabel('Year')
    plt.ylabel('Hurricanes')
    plt.show()
except FileNotFoundError:
    print("File 'hurricane_data.csv' not found.")

# 5. Plot Histogram for San Francisco and Moscow Temperatures
try:
    # Load temperature data
    temperature_data = pd.read_csv('city_temperature.csv')
    sf_temps = temperature_data[temperature_data['City'] == 'San Francisco']['Temperature']
    moscow_temps = temperature_data[temperature_data['City'] == 'Moscow']['Temperature']

    # Plot histograms
    plt.hist(sf_temps, alpha=0.5, label='San Francisco')
    plt.hist(moscow_temps, alpha=0.5, label='Moscow')
    plt.legend()
    plt.title('Temperature Histogram')
    plt.xlabel('Temperature')
    plt.ylabel('Frequency')
    plt.show()
except FileNotFoundError:
    print("File 'city_temperature.csv' not found.")

# 6. Create CSV and Scatter Plot
data_dict = {
    'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
    'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
    'female': [0, 1, 1, 0, 1],
    'age': [42, 52, 36, 24, 73],
    'preTestScore': [4, 24, 31, 2, 3],
    'postTestScore': [25, 94, 57, 62, 70]
}
df = pd.DataFrame(data_dict)

# Scatter Plot with size determined by 'age'
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['age'] * 10, alpha=0.5)
plt.title('Scatter Plot of PreTestScore vs PostTestScore')
plt.xlabel('PreTestScore')
plt.ylabel('PostTestScore')
plt.show()

# Scatter Plot with color determined by 'female'
plt.scatter(df['preTestScore'], df['postTestScore'], s=300, c=df['female'], alpha=0.5, cmap='coolwarm')
plt.title('Scatter Plot with Color by Gender')
plt.xlabel('PreTestScore')
plt.ylabel('PostTestScore')
plt.show()
```
