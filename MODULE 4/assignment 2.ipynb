```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
math_scores = pd.read_csv('MathScoreTerm1.csv')
ds_scores = pd.read_csv('DSScoreTerm1.csv')
physics_scores = pd.read_csv('PhysicsScoreTerm1.csv')
```

```python
math_scores.drop(['Name'], axis=1, inplace=True)
ds_scores.drop(['Name'], axis=1, inplace=True)
physics_scores.drop(['Name'], axis=1, inplace=True)
```

```python
columns_to_drop = ['Name', 'Ethnicity']

for col in columns_to_drop:
    if col in math_scores.columns:
        math_scores.drop([col], axis=1, inplace=True)
    if col in ds_scores.columns:
        ds_scores.drop([col], axis=1, inplace=True)
    if col in physics_scores.columns:
        physics_scores.drop([col], axis=1, inplace=True)
```

```python
math_scores.fillna(0, inplace=True)
ds_scores.fillna(0, inplace=True)
physics_scores.fillna(0, inplace=True)
```

```python
final_scores = pd.merge(math_scores, ds_scores, on='ID')
final_scores = pd.merge(final_scores, physics_scores, on='ID')
```

```python
final_scores['Sex'] = final_scores['Sex'].map({'M': 1, 'F': 2})
```

```python
final_scores.to_csv('ScoreFinal.csv', index=False)
print("Merged and transformed data saved to 'ScoreFinal.csv'.")
```
