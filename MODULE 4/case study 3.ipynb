```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Task 1: Plot Bar Graph of Hurricanes by Year
def plot_hurricanes():
    data = pd.read_csv('Hurricanes.csv')  # Replace with the actual dataset path
    plt.figure(figsize=(10, 6))
    plt.bar(data['Year'], data['Hurricanes'], color='skyblue')
    plt.title('Hurricanes by Year', fontsize=16)
    plt.xlabel('Year', fontsize=14)
    plt.ylabel('Number of Hurricanes', fontsize=14)
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Task 2: Plot Histogram of Temperatures for San Francisco and Moscow
def plot_temperatures():
    data = pd.read_csv('CityTemps.csv')  # Replace with the actual dataset path
    sf_temp = data[data['San Francisco'] == 'San Francisco']['San Francisco']
    moscow_temp = data[data['Moscow'] == 'Moscow']['Moscow']

    plt.figure(figsize=(10, 6))
    plt.hist(sf_temp, bins=20, alpha=0.7, label='San Francisco', color='blue')
    plt.hist(moscow_temp, bins=20, alpha=0.7, label='Moscow', color='red')
    plt.title('Temperature Distribution (2014-2015)', fontsize=16)
    plt.xlabel('Temperature', fontsize=14)
    plt.ylabel('Frequency', fontsize=14)
    plt.legend()
    plt.grid(alpha=0.7)
    plt.show()

# Task 3: Create and Read CSV File
def create_and_read_csv():
    data = {
        'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
        'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
        'female': [0, 1, 1, 0, 1],
        'age': [42, 52, 36, 24, 73],
        'preTestScore': [4, 24, 31, 2, 3],
        'postTestScore': [25, 94, 57, 62, 70]
    }
    df = pd.DataFrame(data)
    df.to_csv('output.csv', index=False)  # Save to CSV
    df_loaded = pd.read_csv('output.csv')  # Read CSV
    print(df_loaded)

# Task 4: Create Plots (Simple Plot and Scatter Plot)
def create_plots():
    x = [1, 2, 3, 4]
    y = [20, 21, 20.5, 20.8]
    y_error = [0.12, 0.13, 0.2, 0.1]

    # Simple Plot
    plt.figure(figsize=(4, 5), dpi=100)
    plt.errorbar(x, y, yerr=y_error, fmt='-o', label='Values')
    plt.title('Simple Plot', fontsize=14)
    plt.xlabel('X-axis', fontsize=12)
    plt.ylabel('Y-axis', fontsize=12)
    plt.legend()
    plt.grid()
    plt.show()

    # Scatter Plot
    random_x = np.random.rand(50)
    random_y = np.random.rand(50)
    plt.figure(figsize=(6, 6))
    plt.scatter(random_x, random_y, c=random_y, cmap='viridis', s=100, alpha=0.7, edgecolors='k')
    plt.title('Scatter Plot', fontsize=14)
    plt.xlabel('X-axis', fontsize=12)
    plt.ylabel('Y-axis', fontsize=12)
    plt.colorbar(label='Color Intensity')
    plt.grid()
    plt.show()

# Run Case Study 3 Tasks
plot_hurricanes()
plot_temperatures()
create_and_read_csv()
create_plots()
```
