```python
### Data Manipulation

data = pd.read_csv('Student.csv')
data
```

```python
# Filter Conditon 
# I want to read all the students where age is greather than 20 
# SQL: SELET * FROM data where age >= 20;
data.loc[data.age >= 20]
```

```python
loc() = Based on the location values want to extract the data 
iloc()  = Based on the index value want to extract the data 
```

```python
# I want to read all the students where age is greather than 20 and all Male student 
# SQL: SELET * FROM data where age >= 20 and gender = 'M'; 
data.loc[(data.age >= 20) & (data.gender =='M')]
```

```python
# SQL: SELET city  section FROM data where age >= 20;

data.loc[(data.age >= 20),['city','section']]
```

```python
data
```

```python
data.loc[1:3]
```

```python
data.iloc[1:3]  # n-1 
```

```python
data
```

```python
# Update Operations 
# SQL UPDATE data set column1 = new_Value, column2= new_Value WHERE <Condition>;

data.loc[(data.age >= 20),['section']] = 'M'
data
```

```python
data.loc[(data.age >= 20),['section','city']] = ['Z','Pune']
data
```

```python
# order the data 
# SQL SELECT * FROM data ORDER BY age desc;
data.sort_values(by ='age',ascending=False)
```

```python
# Group by function
# SQL SELECT section , count(*) from data group by section;
data['section'].value_counts()
```

# Pandas Join operations 



```python
import numpy as np
left_df = pd.DataFrame({'join_keys':['a','b','c','d'],'values':np.random.randint(10,99,4)})
right_df = pd.DataFrame({'join_keys':['b','d','e','f'],'values':np.random.randint(10,99,4)})
```

```python
left_df
```

```python
right_df
```

```python
# SQL SELECT * FROm TBL1 t1 JOIN TBL2 t2 on t1.columnname = t2.columnname
left_df.merge(right =right_df, how = 'inner', on ='join_keys') 
```

```python
left_df.merge(right =right_df, how = 'inner', on ='join_keys') 
```

```python
left_df.merge(right =right_df, how = 'left', on ='join_keys') 
```

```python
left_df.merge(right =right_df, how = 'right', on ='join_keys') 
```

```python
left_df.merge(right =right_df, how = 'outer', on ='join_keys') 
```
