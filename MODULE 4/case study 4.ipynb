```python
def plot_sales_trends():
    data = pd.read_csv('data/BigMartSalesData.csv')  # Replace with the actual dataset path
    data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'])
    data['Year'] = data['InvoiceDate'].dt.year
    data['Month'] = data['InvoiceDate'].dt.month

    sales_2011 = data[data['Year'] == 2011]
    monthly_sales = sales_2011.groupby('Month')['SalesAmount'].sum()

    # Line Plot
    plt.figure(figsize=(10, 6))
    plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='green', label='Sales')
    plt.title('Monthly Sales Trend (2011)', fontsize=16)
    plt.xlabel('Month', fontsize=14)
    plt.ylabel('Total Sales', fontsize=14)
    plt.xticks(range(1, 13))
    plt.grid(axis='both', linestyle='--', alpha=0.7)
    plt.legend()
    plt.show()

    # Bar Chart
    plt.figure(figsize=(10, 6))
    plt.bar(monthly_sales.index, monthly_sales.values, color='blue')
    plt.title('Monthly Sales Bar Chart (2011)', fontsize=16)
    plt.xlabel('Month', fontsize=14)
    plt.ylabel('Total Sales', fontsize=14)
    plt.xticks(range(1, 13))
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Pie Chart for Country-wise Sales
def plot_pie_chart():
    data = pd.read_csv('data/BigMartSalesData.csv')
    country_sales = data.groupby('Country')['SalesAmount'].sum()

    plt.figure(figsize=(8, 8))
    plt.pie(country_sales, labels=country_sales.index, autopct='%1.1f%%', shadow=True, startangle=90)
    plt.title('Country-wise Sales Distribution (2011)', fontsize=16)
    plt.show()

# Scatter Plot for Invoice Amounts
def plot_scatter():
    data = pd.read_csv('data/BigMartSalesData.csv')
    invoice_amounts = data['SalesAmount']

    plt.figure(figsize=(10, 6))
    plt.scatter(range(len(invoice_amounts)), invoice_amounts, alpha=0.5, color='red', edgecolors='k')
    plt.title('Invoice Amount Concentration', fontsize=16)
    plt.xlabel('Invoice Index', fontsize=14)
    plt.ylabel('Sales Amount', fontsize=14)
    plt.grid(alpha=0.7)
    plt.show()

# Run Case Study 4 Tasks
plot_sales_trends()
plot_pie_chart()
plot_scatter()
```
