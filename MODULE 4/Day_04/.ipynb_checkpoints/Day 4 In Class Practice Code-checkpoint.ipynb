### Use case - Banking System Operations 

```python
# Parent Class
class User:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
        print('Parent Constructor Created')
    def show_details(self):
        print('Personal Information')
        print('Name = ', self.name )
        print('Age = ', self.age )
        print('Gender = ', self.gender )

# Child Class - Inheritance 
class Bank(User) :
    def __init__(self,name, age, gender):  # By default constructor alone wont get inherited 
        super().__init__(name, age, gender)  # parent class constructor inherited to my child class 
        self.balance = 0 
        print('Child Constructor Created')

    def deposit(self, amount):
        self.amount = amount
        self.balance += self.amount
        print('Account Balance has been updated =',self.balance)
        
    def withdraw(self, amount):
        self.amount = amount
        if self.amount  > self.balance:
            print('Insufficient Fund in your account =',self.balance)
        else:
            self.balance -= self.amount
            print('Account Balance has been updated =',self.balance)
    def view_balance(self):
        self.show_details()
        print('Account Balance =', self.balance)         
```

```python
import math, random

print('1 for Numeric OTP')
print('2 for Apha numeric OTP')

opt_type = int(input('Enter Number to Select OTP Type: '))

if opt_type  == 1 :
    # for nuemeric OTP only
    corpus= "0123456789"   
else:
     # for alpha nuemeric OTP 
    corpus= "0123456789abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ"

generate_OTP = "" 
size=6
length = len(corpus) 
for i in range(size) : 
  generate_OTP+= corpus[math.floor(random.random() * length)] 
print(generate_OTP)
```
