```python
import re

def check_password_validity(password):
    # Define the criteria regex patterns
    criteria = [
        (re.compile(r"[a-z]"), "At least 1 letter between [a-z]"),
        (re.compile(r"[0-9]"), "At least 1 number between [0-9]"),
        (re.compile(r"[A-Z]"), "At least 1 letter between [A-Z]"),
        (re.compile(r"[$#@]"), "At least 1 character from [$#@]"),
    ]

    # Check password length
    if len(password) < 6:
        return "Password must be at least 6 characters long."
    elif len(password) > 12:
        return "Password must be no more than 12 characters long."

    # Validate the password against all criteria
    for pattern, error_message in criteria:
        if not pattern.search(password):
            return error_message

    return "Password is valid."

# Get input from the user
password = input("Enter your password: ")
result = check_password_validity(password)
print(result)
```

```python
import re

def check_password_validity(password):
    if not (6 <= len(password) <= 12):
        return "Password must be 6-12 characters long."

    patterns = [r"[a-z]", r"[0-9]", r"[A-Z]", r"[$#@]"]
    messages = ["At least 1 letter between [a-z]", "At least 1 number between [0-9]", "At least 1 letter between [A-Z]", "At least 1 character from [$#@]"]

    for pattern, message in zip(patterns, messages):
        if not re.search(pattern, password):
            return message

    return "Password is valid."

password = input("Enter your password: ")
print(check_password_validity(password))
```

```python
# Accept input from the user
input_string = input("Enter a string: ")

# Create a dictionary to count occurrences of each character
char_count = {}

for char in input_string:
    char_count[char] = char_count.get(char, 0) + 1

# Print the character counts
for char, count in char_count.items():
    print(f"'{char}': {count}")
```

```python
# Given list
original_list = [12, 24, 35, 24, 88, 120, 155]

# Remove the value 24 using list comprehension
filtered_list = [x for x in original_list if x != 24]

# Print the resulting list
print("List after removing 24:", filtered_list)

```

```python
# Given list
original_list = [12, 24, 35, 70, 88, 120, 155]

# Use list comprehension to filter out the elements at indices 0, 4, and 5
filtered_list = [x for i, x in enumerate(original_list) if i not in (0, 4, 5)]

# Print the resulting list
print("List after removing 0th, 4th, and 5th numbers:", filtered_list)
```
