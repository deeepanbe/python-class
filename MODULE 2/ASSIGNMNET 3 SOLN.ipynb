```python
import base64

# Function to validate the Reference ID
def validate_reference_id(reference_id):
    # Check if the length is 12 and contains alphanumeric characters or special characters
    if len(reference_id) == 12 and reference_id.isalnum():
        return True
    else:
        print("Invalid Reference ID. Please make sure it is 12 characters long and contains only alphanumeric characters.")
        return False

# Function to encrypt the Reference ID using Base64 encoding
def encrypt_reference_id(reference_id):
    encoded_id = base64.b64encode(reference_id.encode('utf-8'))
    return encoded_id.decode('utf-8')

# Function to decrypt the Reference ID
def decrypt_reference_id(encrypted_id):
    decoded_id = base64.b64decode(encrypted_id.encode('utf-8'))
    return decoded_id.decode('utf-8')

# Main function to handle the user input and encryption process
def main():
    reference_id = input("Enter the 12-digit Reference ID: ")
    
    # Validate the Reference ID
    if validate_reference_id(reference_id):
        # Encrypt the Reference ID
        encrypted_id = encrypt_reference_id(reference_id)
        print(f"Encrypted Reference ID: {encrypted_id}")
        
        # Ask if the user wants to decrypt the ID
        decrypt_choice = input("Do you want to decrypt the Reference ID? (y/n): ").lower()
        
        if decrypt_choice == 'y':
            decrypted_id = decrypt_reference_id(encrypted_id)
            print(f"Decrypted Reference ID: {decrypted_id}")
        else:
            print("You chose not to decrypt the Reference ID.")

# Run the program
if __name__ == "__main__":
    main()
```
